```python
import matplotlib.pyplot as plt
import networkx as nx
```

```python
n = 10  # number of nodes
m = 40  # number of edges
seed = 42  # random seed for reproducibility

# Generate an Erdős-Rényi graph
G = nx.erdos_renyi_graph(n, m / (n * (n - 1) / 2), seed=seed)
plt.figure(figsize=(8, 6))
nx.draw(G, with_labels=True)
```

```python
n = 10  # number of nodes
m = 20  # number of edges
seed = 42  # random seed for reproducibility

# Generate an Erdős-Rényi graph
G = nx.gnm_random_graph(n, m, seed=seed)

#some properties of the graph
print("Number of nodes:", G.number_of_nodes())
print("Number of edges:", G.number_of_edges())

print("node degree clustering")
for v in nx.nodes(G):
    print(f"Node {v}: degree={G.degree(v)}, clustering={nx.clustering(G, v)}")
    
print()
print("the adjacency list of the graph")
for line in nx.generate_adjlist(G):
    print(line)
    
pos = nx.spring_layout(G)  # positions for all nodes
nx.draw(G, pos, with_labels=True, node_color='lightblue', edge_color='gray', node_size=500, font_size=10)
plt.show()

# plt.figure(figsize=(8, 6))
# nx.draw(G, with_labels=True)
```
